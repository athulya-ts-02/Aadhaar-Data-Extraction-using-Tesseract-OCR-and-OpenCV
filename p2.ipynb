```python
img = cv2.imread(r"F:\74027a24-38b2-4c87-9ced-ab74615699e2 (1).jpeg")
cv2.imshow("Original", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
grey=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
(thrsh,img_bin)=cv2.threshold(grey,200,255,cv2.THRESH_BINARY | cv2.THRESH_OTSU)
```

```python
cv2.imshow("img",img_bin)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
data=pytesseract.image_to_data(img_bin,output_type=pytesseract.Output.DICT)
```

```python
print(data['text'])
```

```python
full_text = pytesseract.image_to_string(img_bin)
print(full_text)
```

```python
aadhaar_match = re.search(r'\b\d{4}\s\d{4}\s\d{4}\b', full_text)
aadhaar_no    = aadhaar_match.group(0) if aadhaar_match else "Not Found"
print("Aadhaar No :", aadhaar_no)
```

```python
name = "Not Found"
lines = full_text.splitlines()
lines = [l.strip() for l in lines if l.strip()]

for idx, line in enumerate(lines):
    if re.search(r'Date of Birth|DOB', line, re.IGNORECASE):
        for prev in range(idx-1, -1, -1):
            prev_line = lines[prev].strip()
            if re.match(r'^[A-Za-z\s]{3,40}$', prev_line):
                name = prev_line
                break
        break

print("Name       :", name)
```

```python
dob = "Not Found"

for i in range(len(data['text'])):
    if re.search(r'\d{1,2}[\/\-]\d{1,2}[\/\-]\d{4}', data['text'][i]):
        dob = data['text'][i]
        break

print("DOB        :", dob)
```

```python
img = cv2.imread(r"F:\74027a24-38b2-4c87-9ced-ab74615699e2 (1).jpeg")

for i in range(len(data['text'])):
    if data['conf'][i] > 10:
        word = data['text'][i].strip()
        # Match exactly 4 digits that are NOT a year
        if re.search(r'^\d{4}$', word):
            if not re.search(r'^(19|20)\d{2}$', word):
                (x, y, w, h) = (data['left'][i], data['top'][i],
                                 data['width'][i], data['height'][i])
                img = cv2.rectangle(img, (x,y), (x+w,y+h), (0,255,0), 2)
                cv2.putText(img, "Aadhaar", (x,y-5),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0), 1)

cv2.imshow("aadhaar", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
